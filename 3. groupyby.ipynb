```python
# Criando a conexão
import sqlite3
con = sqlite3.connect("BaseDados.db")
```

```python
# Criando um cursos
cur = con.cursor()
```

```python
# Buscando todas as informações da tabela
cur.execute('SELECT * FROM dados')

resultado = cur.fetchall()
```

```python
import pandas as pd

resultado = pd.DataFrame(resultado)

resultado.columns =  [
    i[0] for i in cur.description
]

resultado.head()
```

```python
# Função para consultar dados

def executa_sql(comando):
    cur.execute(comando)
    resultado = cur.fetchall()
    resultado = pd.DataFrame(resultado)
    resultado.columns = [i[0] for i in cur.description]
    print(resultado.shape)
    #display(resultado.head())

    return resultado
```

```python
executa_sql('SELECT MIN(dias_ultimo_acesso) as Minimo, \
            MAX(dias_ultimo_acesso) as Máximo, \
            AVG(dias_ultimo_acesso) as Média FROM dados')
```

```python
executa_sql('SELECT nome_aluno, COUNT(id_aluno) FROM dados GROUP BY nome_aluno')
```

```python
executa_sql('SELECT  nome_aluno, AVG(id_aluno) as Média FROM dados WHERE nota_prova is not null GROUP BY nome_aluno')
```

```python
executa_sql('SELECT nome_aluno, AVG(id_aluno) as Média FROM dados \
            WHERE nota_prova is not null \
            GROUP BY nome_aluno \
            ORDER BY AVG(id_aluno) desc')
```
