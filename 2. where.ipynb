```python
# Criando a conexão
import sqlite3
con = sqlite3.connect("BaseDados.db")
```

```python
# Criando um cursos
cur = con.cursor()
```

```python
# Buscando todas as informações da tabela
cur.execute('SELECT * FROM dados')

resultado = cur.fetchall()
```

```python
import pandas as pd

resultado = pd.DataFrame(resultado)

resultado.columns =  [
    i[0] for i in cur.description
]

resultado.head()
```

```python
# Função para consultar dados

def executa_sql(comando):
    cur.execute(comando)
    resultado = cur.fetchall()
    resultado = pd.DataFrame(resultado)
    resultado.columns = [i[0] for i in cur.description]
    print(resultado.shape)
    #display(resultado.head())

    return resultado
```

```python
executa_sql('SELECT * FROM dados WHERE acesso_plataforma = 0')
```

```python
executa_sql('SELECT DISTINCT nome_aluno, cod_matricula, [e-mail] as email FROM dados \
             WHERE acesso_plataforma = 0 AND dias_ultimo_acesso <= 2') 
```

```python
executa_sql('SELECT DISTINCT nome_aluno, cod_matricula, [e-mail] as email, \
            acesso_plataforma, dias_ultimo_acesso FROM dados \
             WHERE acesso_plataforma = 0 OR dias_ultimo_acesso > 10 ')
```
